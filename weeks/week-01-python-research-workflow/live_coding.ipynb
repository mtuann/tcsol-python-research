## 1. Điểm bắt đầu nghiên cứu của tôi

Chỉ sửa phần giá trị nằm trong dấu ngoặc kép. Giữ nguyên dấu ngoặc kép.

## 2. Đọc một tệp CSV nhỏ

Tệp CSV là một bảng được lưu dưới dạng text. Tuần này, mỗi dòng là một hướng project có thể phát triển.

```python
from pathlib import Path
import csv

possible_paths = [
    Path("data/raw/week01_research_tracks.csv"),
    Path("weeks/week-01-python-research-workflow/data/raw/week01_research_tracks.csv"),
]

data_path = next(path for path in possible_paths if path.exists())
week_dir = data_path.parents[2]

print("Data file exists:", data_path.exists())
print("Data file path:", data_path)
```

```python
with data_path.open(encoding="utf-8") as file:
    rows = list(csv.DictReader(file))

print("Number of rows:", len(rows))
print("Column names:", list(rows[0].keys()))
```

## 3. Xem các dòng dữ liệu

Đừng lo nếu bạn chưa hiểu mọi ký hiệu. Hãy đọc output được in ra và hỏi: một dòng dữ liệu đại diện cho điều gì?

```python
for row in rows:
    print(row["track_id"], "-", row["track"])
    print("  Question:", row["small_research_question"])
    print("  One row in future data:", row["unit_of_observation"])
    print()
```

## 4. Chọn một hướng nghiên cứu

Chọn một trong các ID sau:

- `TCSOL_SHORT`
- `CONTRASTIVE`
- `MTPE`
- `POLICY`

Chỉ sửa giá trị của `selected_track_id`.

```python
selected_track_id = "TCSOL_SHORT"

selected_rows = [row for row in rows if row["track_id"] == selected_track_id]
selected = selected_rows[0]

print("Selected track:", selected["track"])
print("Small research question:", selected["small_research_question"])
print("Data unit:", selected["unit_of_observation"])
print("Likely paper output:", selected["likely_output"])
```

## 5. Tạo một bảng tóm tắt nhỏ

Bảng này rất đơn giản, nhưng đã có thói quen viết paper: nối câu hỏi, dữ liệu, và output.

```python
summary = {
    "selected_track": selected["track"],
    "small_research_question": selected["small_research_question"],
    "unit_of_observation": selected["unit_of_observation"],
    "likely_output": selected["likely_output"],
}

for key, value in summary.items():
    print(key + ":", value)
```

```python
output_dir = week_dir / "outputs/tables"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "week01_selected_track_summary.csv"

with output_path.open("w", encoding="utf-8", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=list(summary.keys()))
    writer.writeheader()
    writer.writerow(summary)

print("Saved summary table to:", output_path)
```
