## 2. Đọc một tệp CSV nhỏ

Tệp CSV là một bảng được lưu dưới dạng text. Tuần này, mỗi dòng là một hướng project có thể phát triển.

**Code mẫu - chỉ cần chạy:** cell tiếp theo tìm file CSV. Nếu mở bằng Colab và chưa có file local, Python sẽ tự tải bản public từ GitHub.

```python
from pathlib import Path
from urllib.request import urlretrieve
import csv

possible_paths = [
    Path("data/raw/week01_research_tracks.csv"),
    Path("weeks/week-01-python-research-workflow/data/raw/week01_research_tracks.csv"),
]

data_path = next((path for path in possible_paths if path.exists()), None)

if data_path is None:
    week_dir = Path(".")
    data_dir = week_dir / "data/raw"
    data_dir.mkdir(parents=True, exist_ok=True)
    data_path = data_dir / "week01_research_tracks.csv"
    source_url = "https://raw.githubusercontent.com/mtuann/tcsol-python-research/main/weeks/week-01-python-research-workflow/data/raw/week01_research_tracks.csv"
    try:
        urlretrieve(source_url, data_path)
    except Exception as error:
        raise RuntimeError(
            "Python could not download the Week 01 CSV. "
            "Please upload week01_research_tracks.csv to data/raw/ and run this cell again."
        ) from error
else:
    week_dir = data_path.parents[2]

print("Data file exists:", data_path.exists())
print("Data file path:", data_path)
```

```python
with data_path.open(encoding="utf-8") as file:
    rows = list(csv.DictReader(file))

print("Number of rows:", len(rows))
print("Column names:", list(rows[0].keys()))
```

## 3. Xem các dòng dữ liệu

**Code mẫu - chỉ cần chạy:** đừng lo nếu bạn chưa hiểu mọi ký hiệu. Hãy đọc output được in ra và hỏi: một dòng dữ liệu đại diện cho điều gì?

```python
for row in rows:
    print(row["track_id"], "-", row["track"])
    print("  Question:", row["small_research_question"])
    print("  One row in future data:", row["unit_of_observation"])
    print()
```

## 4. Chọn một hướng nghiên cứu

**Bạn sửa:** chọn một trong các ID sau, rồi chỉ sửa giá trị của `selected_track_id`.

- `TCSOL_SHORT`
- `CONTRASTIVE`
- `MTPE`
- `POLICY`

```python
selected_track_id = "TCSOL_SHORT"

valid_track_ids = [row["track_id"] for row in rows]
if selected_track_id not in valid_track_ids:
    raise ValueError("Please choose one of these track IDs: " + ", ".join(valid_track_ids))

selected = next(row for row in rows if row["track_id"] == selected_track_id)

print("Selected track:", selected["track"])
print("Small research question:", selected["small_research_question"])
print("Data unit:", selected["unit_of_observation"])
print("Likely paper output:", selected["likely_output"])
```

## 5. Tạo một bảng tóm tắt nhỏ

**Code mẫu - chỉ cần chạy:** bảng này rất đơn giản, nhưng đã có thói quen viết paper: nối câu hỏi, dữ liệu, và output.

```python
summary = {
    "selected_track": selected["track"],
    "small_research_question": selected["small_research_question"],
    "unit_of_observation": selected["unit_of_observation"],
    "likely_output": selected["likely_output"],
}

for key, value in summary.items():
    print(key + ":", value)
```

```python
output_dir = week_dir / "outputs/tables"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "week01_selected_track_summary.csv"

with output_path.open("w", encoding="utf-8", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=list(summary.keys()), lineterminator="\n")
    writer.writeheader()
    writer.writerow(summary)

print("Saved summary table to:", output_path)
```
